```python
import sqlite3
import pandas as pd

conexion = sqlite3.connect("rrhh.db")
df = pd.read_sql("SELECT * FROM empleados", conexion)
conexion.close()

print(df)
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
porcentaje = (df.isnull().sum() / len(df) * 100).round(1)
print(porcentaje)
```

```python
df_sin_nulos = df.dropna()
print(f"Filas originales: {len(df)}")
print(f"Filas sin nulos:  {len(df_sin_nulos)}")

```

```python
# Solo eliminar si el salario es nulo — esas filas son inutilizables para análisis salarial
df_sin_salario_nulo = df.dropna(subset=["salario"])
print(len(df_sin_salario_nulo))
```

```python
# Rellenar ciudad desconocida con un texto explícito
df["ciudad"] = df["ciudad"].fillna("Sin datos")

# Rellenar departamento con el más común (moda)
departamento_mas_comun = df["departamento"].value_counts().iloc[0]
df["departamento"] = df["departamento"].fillna(departamento_mas_comun)
```

```python
#Ejercicio 1 
import sqlite3
import pandas as pd

conexion = sqlite3.connect("./practica/medicare.db")
#tablas que existen 
df_medicare = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conexion
)
print(df_medicare)
```

```python
df_pacientes = pd.read_sql(
    "SELECT * FROM pacientes",
    conexion
)
print(df_pacientes)
```

```python
df_turnos = pd.read_sql(
    "SELECT * FROM turnos",
    conexion
)
print(df_turnos)
```

```python
df_medicos = pd.read_sql(
    "SELECT * FROM medicos",
    conexion
)
print(df_medicos)
```

```python
print(f"forma de tabla pacientes \n", df_pacientes.shape)
print(f"forma de tabla medicos \n", df_medicos.shape)
print(f"forma de tabla turnos \n", df_turnos.shape)
```

```python
#Ejercicio 2
print(f"Nulos de pacientes \n", df_pacientes.isnull().sum())
print(f"Nulos de medicos \n", df_medicos.isnull().sum())
print(f"Nulos de turnos \n", df_turnos.isnull().sum())

```

```python
tablas = {
    "Pacientes": df_pacientes,
    "Medicos": df_medicos,
    "Turnos": df_turnos
}

for nombre, df in tablas.items():
    porcentaje = (df.isnull().sum() / len(df) * 100).round(1)
    print(f"Porcentaje de nulos de {nombre}:\n{porcentaje}\n")
```

```python
#ver filas duplicadas
# Buscar filas iguales ignorando el id
for nombre, tabla in {
    "Pacientes": df_pacientes,
    "Médicos": df_medicos,
    "Turnos": df_turnos
}.items():
    duplicados = tabla[tabla.drop(columns="id").duplicated(keep=False)]
    print(f"\nDuplicados en {nombre}:")
    print(duplicados)
```
