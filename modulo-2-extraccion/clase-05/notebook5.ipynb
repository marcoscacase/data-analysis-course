```python
import pandas as pd
```

```python
#SERIES Y PANDAS

ventas = [
    {"producto": "Notebook",  "categoria": "tech",    "precio": 320000},
    {"producto": "Mouse",     "categoria": "tech",    "precio": 8500},
    {"producto": "Escritorio","categoria": "muebles", "precio": 85000},
    {"producto": "Silla",     "categoria": "muebles", "precio": 45000},
]

totales = {}
for v in ventas:
    cat = v["categoria"]
    if cat not in totales:
        totales[cat] = 0
    totales[cat] += v["precio"]

print(totales)
# {'tech': 328500, 'muebles': 130000}
```

```python
#Ejercicio 4 clase 5

import pandas as pd
 
productos = pd.DataFrame([
    {"nombre": "Notebook", "precio": 320000, "stock": 5},
    {"nombre": "Mouse",    "precio": 8500,   "stock": 42},
    {"nombre": "Teclado",  "precio": 45000,  "stock": 12},
])
 
resultado_a = productos["precio"]
resultado_b = productos[["precio"]]
resultado_c = productos[["nombre", "precio"]]
 
print(type(resultado_a))
print(type(resultado_b))
print(type(resultado_c))
#el primer resultado es una serie porque solo seleccionamos una columna, mientras que los otros dos resultados son dataframes porque seleccionamos una o más columnas dentro de una lista.
#la serie es un objeto unidimensional que puede contener datos de diferentes tipos, mientras que el dataframe es un objeto bidimensional que puede contener múltiples series de datos.
```

```python
#iloc y loc para seleccionar filas y columnas de un DataFrame.
#loc es para selecciona filas y columnas por etiquetas
print(ventas.loc[[1,2] ,["producto", "precio"]])
```

```python
#iloc es para seleccionar filas y columnas por posición

print(ventas.iloc[2:4, [0, 2]])
```

```python
#Ejercicio 5 clase 5
import pandas as pd
 
notas = pd.DataFrame([
    {"alumno": "Ana",    "nota": 8.5},
    {"alumno": "Luis",   "nota": 6.0},
    {"alumno": "Clara",  "nota": 9.2},
    {"alumno": "Pedro",  "nota": 5.5},
    {"alumno": "María",  "nota": 7.8},
    {"alumno": "Jorge",  "nota": 4.0},
])
 
# 1. Muestra solo la primera fila
primera_fila = notas.iloc[0:1]
 
# 2. Muestra las filas en las posiciones 2, 3 y 4 (Clara, Pedro, María)
rango_medio = notas.iloc[2:5]
 
# 3. Muestra solo la última fila
# Pista: pensá en los índices negativos que usaste con listas en clases anteriores
ultima_fila = notas.iloc[-1:]
 
print(primera_fila)
print(rango_medio)
print(ultima_fila)
```
