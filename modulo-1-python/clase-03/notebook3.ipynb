```python
#Ejercicio 2 clase 3

def calcular_orden(precio_unitario, cantidad, descuento):
    subtotal = precio_unitario * cantidad
    total = subtotal * (1 - descuento)
    return subtotal, descuento, total

#Caso 1: 3 unidades a $12.000, sin descuento
#Caso 2: 5 unidades a $8.500, 10% descuento
#Caso 3: 2 unidades a $45.000, 25% descuento

#Ejemplo caso 2:
#subtotal: $42.500 | Descuento: $4.250 | Total: $38.250

total1 = calcular_orden(12000, 3, 0)
total2 = calcular_orden(8500, 5, 0.10)
total3 = calcular_orden(45000, 2, 0.25)

print(f"Caso 1: Subtotal: ${total1[0]:,.2f} | Descuento: ${total1[1]*total1[0]:,.2f} | Total: ${total1[2]:,.2f}")
print(f"Caso 2: Subtotal: ${total2[0]:,.2f} | Descuento: ${total2[1]*total2[0]:,.2f} | Total: ${total2[2]:,.2f}")
print(f"Caso 3: Subtotal: ${total3[0]:,.2f} | Descuento: ${total3[1]*total3[0]:,.2f} | Total: ${total3[2]:,.2f}")
```

```python
def convertir_temperatura(registro):
    try:
        celsius = float(registro["valor"])
    except ValueError:
        print(f"sin lectura para el sensor {registro['id']}")
        return None
    return (celsius * 9/5) + 32
 
sensores = [
    {"id": "S-01", "valor": "87.3"},
    {"id": "S-02", "valor": "N/A"},
    {"id": "S-03", "valor": "92.1"},
    {"id": "S-04", "valor": ""},
    {"id": "S-05", "valor": "88.7"},
]
 
for sensor in sensores:
    fahrenheit = convertir_temperatura(sensor)
    if fahrenheit is not None:
        print(f"Sensor {sensor['id']}: {fahrenheit:.1f} °F")
```

```python
#Ejercicio 4 clase 3
registros = [
    {"producto": "Notebook",  "precio": "320000", "stock": "5",  "categoria": "tech"},
    {"producto": "Escritorio","precio": "85000",  "stock": "N/A","categoria": "muebles"},
    {"producto": "Mouse",     "precio": "-8500",  "stock": "42", "categoria": "tech"},
    {"producto": "",          "precio": "45000",  "stock": "8",  "categoria": "muebles"},
    {"producto": "Monitor",   "precio": "210000", "stock": "0",  "categoria": "tech"},
    {"producto": "Auriculares","precio": "32000", "stock": "11", "categoria": "tech"},
]
 
resultado = []
errores = []
 
for r in registros:
    try:
        if r["producto"] == "":
            raise ValueError("nombre vacío")
        precio = float(r["precio"])
        if precio <= 0:
            raise ValueError(f"precio inválido: {r['precio']}")
        stock = int(r["stock"])
        valor_inventario = precio * stock
        resultado.append({
            "producto": r["producto"],
            "precio": precio,
            "stock": stock,
            "valor": valor_inventario,
            "categoria": r["categoria"],
        })
    except (ValueError, KeyError) as e:
        errores.append({"producto": r.get("producto", "?"), "error": str(e)})
 
tech    = [r for r in resultado if r["categoria"] == "tech"]
muebles = [r for r in resultado if r["categoria"] == "muebles"]
 
print(f"Procesados: {len(resultado)} | Errores: {len(errores)}")
print(f"\nTech    — valor total: ${sum(r['valor'] for r in tech):,}")
print(f"Muebles — valor total: ${sum(r['valor'] for r in muebles):,}")
 
for e in errores:
    print(f"  ✗ {e['producto']}: {e['error']}")

#refaccionando el codigo de arriba en funciones
```
