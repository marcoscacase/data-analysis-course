List comprehension (resumir un for en 1 linea, maximo 1 if porque sino es engorroso y no se entiende)

```python
precios = [8500, 32000, 15900, 7200, 54000, 12300]

resultado = []
for p in precios:
    if p > 10000:
        resultado.append(p * 1.19)

print(resultado)

print("Resultado usando comprensión de listas:")

#[ qué quiero hacer con cada elemento   for elemento in la lista original   if condición opcional ]
resultado = [p * 1.19 for p in precios if p > 10000]
print(resultado)

nombres = ["ana", "luis", "carla"]
capitalizados = [n.capitalize() for n in nombres]
print(capitalizados)
# ['Ana', 'Luis', 'Carla']
```

Limpieza de strings 
.strip() (Limpieza de espacios)
.replace() (reemplaza "", "", primero el valor a reemplazar y luego por lo que quiero reemplazar)



```python
sucio = "  $45.000  "

print("valor antes de metodos de limpieza: ", sucio)
limpio = sucio.strip().replace("$", "").replace(".", "")
print("valor despues de metodos de limpieza: ", limpio)
# "45000"
```

```python
def limpiar_precio(valor):
    try:
        limpio = float(valor.strip().replace("$", "").replace(",", "").replace(".", ""))
        return limpio
    except ValueError:
        return None

precios_crudos = ["$45.000", "$1,250.50", " $8500 ", "N/A", "$32.000"]

precios_limpios = [limpiar_precio(p) for p in precios_crudos]
print(precios_limpios)
# [45000.0, 1250.5, 8500.0, None, 32000.0]

precios_validos = [p for p in precios_limpios if p is not None]
print(precios_validos)
# [45000.0, 1250.5, 8500.0, 32000.0]

print("otra manera")
def limpiar_precio(valor):
    try:
        limpio = valor.strip().replace("$", "")

        if "," in limpio and "." in limpio:
            limpio = limpio.replace(",", "")
        elif "." in limpio:
            if len(limpio.split(".")[-1]) == 3:
                limpio = limpio.replace(".", "")
        elif "," in limpio:
            limpio = limpio.replace(",", "")

        return float(limpio)

    except ValueError:
        return None


precios_crudos = ["$45.000", "$1,250.50", " $8500 ", "N/A", "$32.000"]

precios_limpios = [limpiar_precio(p) for p in precios_crudos]
print(precios_limpios)

precios_validos = [p for p in precios_limpios if p is not None]
print(precios_validos)
```

```python
#Ejercicio 5 clase 4
categorias_crudas = ["  tecnologia", "MUEBLES", "Tecnologia ", "muebles", " ROPA", "ropa  "]
 
def normalizar_categoria(categoria):
    return categoria.strip().capitalize()
 
# Tu tarea: normalizar todas las categorías
categorias_normalizadas = [normalizar_categoria(d) for d in categorias_crudas]
 
# Tu tarea: obtener las categorías únicas (pista: investigá la función set())
categorias_unicas = set(categorias_normalizadas)
 
print(categorias_normalizadas)
print(categorias_unicas)
```

```python
#Ejercicio 6 clase 4
precios_usa = ["$1,200.50", "$45.00", "N/A", "$3,890.25", "", "$120.00"]
 
def limpiar_precio_usa(valor):
    try:
        limpio = valor.strip().replace("$", "").replace(",", "")
        return float(limpio)
    except ValueError:
        return None
 
# Tu tarea: aplicar la función a toda la lista
precios_procesados = [limpiar_precio_usa(p) for p in precios_usa]
 
# Tu tarea: quedarte solo con los que no son None
precios_validos = [p for p in precios_procesados if p is not None]
 
print(precios_procesados)
print(precios_validos)
```
