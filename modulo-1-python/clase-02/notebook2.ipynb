Estructuras Condicionales

```python
num1 = 20
num2 = 5

condicional = num1 > num2

if condicional:
    print("num1 es mayor que num2")
else:
    print("num1 no es mayor que num2")
```

Operadores Logicos

```python
email = "marcoscacase@gmail.com"
password = "12345"

if email == "marcoscacase@gmail.com" and password == "12345":
    print("Inicio de sesión exitoso")
else:
    print("Credenciales incorrectas")
```

```python
temperatura = 30.3

if temperatura <= 60:
    estado = "normal"
elif temperatura <= 80:
    estado = "advertencia"
else:
    estado = "crítico"

print(f"Temperatura: {temperatura}°C — Estado: {estado}")
```

Estructuras repetitivas (FOR)

```python
precios = [15000, 32000, 8500, 71000, 12300]
print(precios)
print (precios[0])  # Primer elemento
print (precios[-1])  # Último elemento

print("estructura for")
#espacio de memoria #Lista a procesar
frutas = ["manzana", "banana", "naranja", "kiwi"]
for fruta in frutas:
    print(fruta + "s")
for precio in precios:
    print(f"Precio con IVA: {precio * 1.21}")  # Aplicando un aumento del 21% a cada precio
```
