```python
# Ejercicio 3
# Punto de partida
inventario = {
    "NB-001": {"nombre": "Notebook Pro", "precio": 320000, "stock": 5},
    "MS-002": {"nombre": "Mouse Inalámbrico", "precio": 8500, "stock": 42},
    "MN-003": {"nombre": "Monitor 24\"", "precio": 210000, "stock": 0},
    "TC-004": {"nombre": "Teclado Mecánico", "precio": 45000, "stock": 12},
}

codigo_busqueda = "TC-004"

# Tu tarea: buscar el código en el inventario
# Si existe: imprimir nombre, precio y stock
# Si no existe: imprimir "Producto no encontrado"
#Usa .get() para acceder a los valores del diccionario de manera segura
producto = inventario.get(codigo_busqueda)

if producto:
    print(f"Nombre: {producto['nombre']}")
    print(f"Precio: {producto['precio']}")
    print(f"Stock: {producto['stock']}")
else:
    print("Producto no encontrado")
```

```python
# Punto de partida
ventas = [23000, 87000, 45000, 102000, 31500, 68000, 49000, 95000, 12000, 74000]
meta = 50000

sobre_meta = []
bajo_meta = []

# Tu tarea: clasificar cada venta en el grupo correcto y calcular los totales
for venta in ventas:
    if venta >= meta:
        sobre_meta.append(venta)
    else:
        bajo_meta.append(venta)

total_sobre_meta = sum(sobre_meta)
total_bajo_meta = sum(bajo_meta)

print(f"Ventas sobre la meta: {sobre_meta} — Total: {total_sobre_meta}")
print(f"Ventas bajo la meta: {bajo_meta} — Total: {total_bajo_meta}")
```

```python
# Punto de partida Ejercicio 7
transacciones = [
    {"id": "TX001", "pais": "MX", "monto": 45000,  "moneda": "MXN"},
    {"id": "TX002", "pais": "AR", "monto": 128000, "moneda": "ARS"},
    {"id": "TX003", "pais": "CO", "monto": 0,      "moneda": "COP"},
    {"id": "TX004", "pais": "MX", "monto": 87000,  "moneda": "MXN"},
    {"id": "TX005", "pais": "AR", "monto": -5000,  "moneda": "ARS"},
    {"id": "TX006", "pais": "PE", "monto": 32000,  "moneda": "PEN"},
    {"id": "TX007", "pais": "CO", "monto": 215000, "moneda": "COP"},
    {"id": "TX008", "pais": "MX", "monto": 61000,  "moneda": "MXN"},
    {"id": "TX009", "pais": "AR", "monto": 94000,  "moneda": "ARS"},
    {"id": "TX010", "pais": "PE", "monto": 17000,  "moneda": "PEN"},
]

# Tu tarea: construir el reporte completo
# Pista: para las top 3 investigá sorted() con el parámetro key=
#Total de transacciones procesadas
#Monto total y promedio
#Desglose por pais (cantidad de transacciones y monto total)
#Top 3 transacciones por monto
#Transacciones con monto negativo o cero con su motivo

total_transacciones = len(transacciones)
monto_total = sum(transaccion["monto"] for transaccion in transacciones)
promedio_monto = monto_total / total_transacciones if total_transacciones > 0 else 0
desglose_por_pais = {}
for transaccion in transacciones:
    pais = transaccion["pais"]
    if pais not in desglose_por_pais:
        desglose_por_pais[pais] = {"cantidad": 0, "monto_total": 0}
    desglose_por_pais[pais]["cantidad"] += 1
    desglose_por_pais[pais]["monto_total"] += transaccion["monto"]

# Top 3 transacciones por monto
top_3_transacciones = sorted(transacciones, key=lambda x: x["monto"], reverse=True)[:3]

# Transacciones con monto negativo o cero
transacciones_negativas_cero = [transaccion for transaccion in transacciones if transaccion["monto"] <= 0]

print(f"Total de transacciones procesadas: {total_transacciones}")
print(f"Monto total: {monto_total}")
print(f"Promedio de monto: {promedio_monto}")
print("Desglose por país:")
for pais, datos in desglose_por_pais.items():
    print(f"  {pais}: {datos['cantidad']} transacciones, monto total: {datos['monto_total']}")
print("Top 3 transacciones por monto:")
for i, transaccion in enumerate(top_3_transacciones, start=1):
    print(f"  {i}. {transaccion['id']}: {transaccion['monto']}")
print("Transacciones con monto negativo o cero:")
for transaccion in transacciones_negativas_cero:
    print(f"  {transaccion['id']}: {transaccion['monto']}")
```
