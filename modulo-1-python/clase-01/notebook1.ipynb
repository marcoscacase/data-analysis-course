Operadores artimeticos, relacionales 

```python
numero1 = 20
numero2 = 10
suma = numero1 + numero2
print(suma)
resta = numero1 - numero2
print(resta)
multiplicacion = numero1 * numero2
print(multiplicacion)
division = numero1 / numero2
print(division)

#-------------------------------------------------------------
print (numero1 == numero2)
print (numero1 != numero2)
print (numero1 > numero2)
print (numero1 < numero2)
print (numero1 >= numero2)
print (numero1 <= numero2)
```
